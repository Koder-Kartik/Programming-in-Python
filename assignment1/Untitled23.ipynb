```python
n = int(input('Enter number'))
i=2
while i<n :
  if n%i == 0 :
    print('Not Prime')
    break
  i=i+1

if i==n:
  print('Prime')
```

```python
# @title
for n in range(1, 21):
    if n < 2:
        print(n, "is Not Prime")
        continue

    for i in range(2, n):
        if n % i == 0:
            print(n, "is Not Prime")
            break
    else:
        print(n, "is Prime")
```
